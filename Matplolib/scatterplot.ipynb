```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
df=pd.read_csv('student_depression_dataset.csv') 
df
```

```python
df.head()
```

```python
df['Age'].value_counts()
```

```python
#Age vs Average CGPA
k=df.groupby('Age')['CGPA'].mean().head(10)
k
```

```python
x=k.index
y=k.values
##function to create a scatter plot -->plt.scatter(x,y)
plt.figure(figsize=(10,5))
plt.title("Average CGPA Variration as per Age")
plt.xlabel("Age")
plt.ylabel("Average CGPA")
plt.scatter(x,y,color='red',marker='*')
plt.grid()
plt.show()
```
