```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
df=pd.read_csv('student_depression_dataset.csv') 
df
```

```python
data=df['Gender'].value_counts()
data
```

```python
x=data.index
y=data.values
```

```python
plt.figure(figsize=(8,4))
plt.title("Univariate Analysis on Gender Column")
plt.bar(x,y)
plt.show()
```

```python
df['Dietary Habits'].value_counts()
```

```python
data=df[df['Dietary Habits']!='Others']
data['Dietary Habits'].value_counts()
```

```python
t=data.groupby('Dietary Habits')['Depression'].sum()
t
```

```python
x=t.index
y=t.values
plt.xlabel("Dietary Habits")
plt.ylabel("Number of Peoples")
plt.title("Bar chart of number of people's Dietary habits")
plt.bar(x,y,color="red")
plt.show()
```
