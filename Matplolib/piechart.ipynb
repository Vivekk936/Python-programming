```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
df=pd.read_csv('student_depression_dataset.csv') 
df
```

```python
df.head()
```

```python
#univariate--> Categorical column value data contribution in percentage
#Univariate Analysis -->value_counts()
data=df['Dietary Habits'].value_counts()
data
```

```python
plt.figure(figsize=(10,10))
plt.pie(data,labels=data.index,autopct='%0.2f%%',colors=['yellow','green','magenta','black'],explode=[0.1,0,0,0],shadow=True)
plt.legend()
plt.show()
```
